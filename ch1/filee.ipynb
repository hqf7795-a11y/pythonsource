```python
# fruits.txt 읽어서 단어 개수 출력
# apple 3
# banana 2

# apple banana orange
# 단어 분리 : split() => ['apple','banana','orange']

#dict 추가
# dict1['apple'] += 1
# if "apple" in dict1:
    # dict['apple'] += 1
# else:
    # dict['apple'] = 1

dict2 = {}
with open("./data/fruits.txt","r") as f:
    for line in f:
        # print(line)
        words = line.split()
        # print(words)
        for word in words:
            # print(word)
            if word in dict2:
                dict2[word] += 1
            else:
                dict2[word] = 1

for fruit,count in dict2.items():
    print(f"{fruit}:{count}")
```

```python
# 1000 명의 이름,키,몸무게 info.txt 읽은 후 BMI지수 계산
# BMI = 체중/ 키의 제곱
# BMI가 18.5 미만이면 저체중
# BMI가 18.5~22.9 정상체중
# BMI가 23~24.9 과체중
# BMI가 25 이상 비만
```

```python
with open("./data/info.txt","r",encoding="utf-8") as f:
    for line in f:
        name,weight,height = line.strip().split("," "")
        weight = int(weight)
        height = int(height)
        bmi = weight / (height/100)**2

        if bmi < 18.5:
            result = "저체중"
        elif bmi < 23:
            result = "정상체중"
        elif bmi < 25:
            result = "과체중"
        else:
            result = "비만"
            # print(f"{name} : 몸무게{weight}, 키{height}, bmi : {"%0.1f"%bmi} {result}")
        print("\n".join(["이름: {}", "몸무게 : {}", "키 : {}", "MBI : {}", "결과 : {}"]).format(name,weight,height,"%0.1f"%bmi,result))
        print()
```

```python
# origin.txt 읽어서 암호화 시킨 후 encr.txt 저장
# 1. 암호화 선택 : origin.txt 읽어서 변수에 담기 => 그걸 204처럼
# 2. 복호화 선택 : 암호화된 파일을 읽은 후 원래대로 출력

while True:
    no = int(input("1.암호화 | 2.복호화 | 3.종료 중 원하는 메뉴의 번호를 입력하세요"))

    if no == 1:
        with open("./data/origin.txt","r",encoding="utf-8") as f:
           contents = f.read()

        with open("./data/encr.txt","w",encoding="utf-8") as f:
            for c in contents:
                f.write(chr(ord(c)+100))
    elif no == 2:
        with open("./data/encr.txt","r",encoding="utf-8") as f:
            contents = f.read()
            for c in range(len(contents)):
                print(chr(ord(contents[c]) - 100), end="")
    else:
        break
```
