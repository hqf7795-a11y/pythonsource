```python
# 예외 클래스 생성 가능
class MyError(Exception):
    def __str__(self):
        return "허용되지 않은 별명입니다."

def say_nick(nick):
    if nick == "바보":
        raise MyError()
    print(nick)

try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
```

```python
# 문제 1. 랜덤 비밀번호 생성기

# 사용자로부터 생성할 비밀번호의 길이를 입력받아 랜덤 비밀번호를 생성하는 프로그램 작성
# choice()

chars = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789"
# 비밀번호의 길이는 4 이상만 허용
# 숫자가 아닌 값을 입력하면 예외처리

# choice, shuffle, input,

from random import *

# password = 1234567890



# try:
#     password = int(input("number 입력"))
#     if password != int:
#         print("ss")
# except:
#     print("입력이 정확하지 않습니다.")

# if password < 4:
#     print("too short")
# elif password > 4:
#     choice(shuffle(password))

while True:
    try:
        pwd_num = int(input("num_len : "))

        if pwd_num < 4:
            print("short")
            continue
        break
    except ValueError:
        print("press only num")

# 리스트 컴프리헨션
# password = [choice(chars) for _ in range(pwd_num)]

password = []
for i in range(pwd_num):
    password.append(choice(chars))

# 리스트 => 문자열로 변경
print(f"random password : {"".join(password)}")
```

```python
# 나이 꼐산 프로그램
# 생년월일을 입력받아 현재 나이를 계산하는 프로그램 작성
# 입력형식 2001.08.25
# input() => str로줌 '2001.08.25.'  => 날짜 형식으로 바꾸기

import datetime

# 오늘 날짜
now = datetime.datetime.now()

try:
    # 생년월일 입력
    birth_str = input("press(YYYY-MM-DD)")
    # '2001-05-15'
    birth = datetime.datetime.strptime(birth_str,"%Y-%m-%d")

    #   2026 - 2001
    # now에서 년도만 추출하기
    age = now.year - birth.year
    if (now.month, now.day) < (birth.month,birth.day):
        age -= 1

    print("나이",age)
except:
    print("날짜 형식이 올바르지 않습니다.")
```

```python
# 다음과 같은 csv 파일이 있다.
# 이름,국어,영어,수학
# 길동이,11,22,33
# 가동이,24,22,77
# 나동이,19,55,66
# 다동이,72,42,93
# 라동이,88,41,53

# exlist1 = [
    # {"이름":"가동이에요","국어":88,"영어":92,"수학":33}
    #{"이름":"나동이에요","국어":87,"영어":90,"수학":44}
    # {"이름":"다동이에요","국어":93,"영어":11,"수학":86}
    # {"이름":"라동이에요","국어":87,"영어":66,"수학":97}
    # {"이름":"마동이에요","국어":33,"영어":90,"수학":77}
# ]




import csv

with open("./data/sample11.csv","w",encoding="utf-8",newline="") as f:
    headers = ['이름','국어','영어','수학']
    writer = csv.DictWriter(f,fieldnames=headers)
    
    writer.writeheader()
    writer.writerow({"이름":"가동이에요","국어":88,"영어":92,"수학":33})
    writer.writerow({"이름":"나동이에요","국어":87,"영어":90,"수학":44})
    writer.writerow({"이름":"다동이에요","국어":93,"영어":11,"수학":86})
    writer.writerow({"이름":"라동이에요","국어":87,"영어":66,"수학":97})
    writer.writerow({"이름":"마동이에요","국어":33,"영어":90,"수학":77})

# csv 파일을 읽어 학생별 평균 점수 출력
# 길동이 : 11.0

# csv.DictReader() 사용, 평균은 둘째까지
```

```python
with open("./data/sample11.csv","r",encoding="utf-8") as f:
    reader = csv.DictReader(f)
# print(reader)

    for row in reader:
        print(row)
        avg = (int(row.get('국어')) + int(row.get('영어')) + int(row.get('수학'))) / 3
        print(f"{row.get('이름')} : {avg:.2f}")
```
