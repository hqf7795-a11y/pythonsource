#### read and write file

    파일 열기 모드 | 설명    
    ----------|--------|
    `r` | **읽기(read) 모드** : 파일의 내용을 읽기만 할 때 사용. 파일이 존재하지 않는 경우 오류 발생 |
    `w` | **쓰기(write) 모드** : 파일의 내용을 새로 쓸 때 사용. 기존 파일이 존재하는 경우 내용 삭제 후 새로 작성. 없는 경우 새로 생성 |
    `a` | **추가(append) 모드** : 기존 파일의 마지막에 내용을 추가 할 때 사용. 파일이 없는 경우 새로 생성 |

```python
list1 = ["Kim","Park","Cho"]

# 리스트를 파일로 작성 => writelines()
# str => write()

with open("./data/test1.text","w",encoding="utf-8") as f:
    f.writelines(list1)
```

```python
dict = {"name":"hong","age":25,"addr":"서울"}
with open("./data/test1.text","w",encoding="utf-8") as f:
    for k,v in dict.items():
        f.writelines(f"{k}:{v}\n")
```
