#### 문자열 포매팅 : basic 33
- 문자열 안에 어떤 값을 삽입하는 방법
-
 코드 | 설명 |
 --- | ---|
 `%s` | 문자열(string) |
 `%d` | 정수(integer) |
 `%c` | 문자 1개(character) |
 `%f` | 부동 소수 (floating-point) |
 `%o` | 8진수 (floating-point) |
 `%x` | 16진수 |
 `%%` | % 문자 자체 |

 - a = "호세 %s 잔" % "five"
  print(a) == 호세 five 잔
 - b = "레미마틴 %d잔" % 5
  print(b,sep="") == 레미마틴 5잔
 - c = "%c is my favorit" %"U"
  print(c) == U is my favorit


# hi 문자열 출력시 10개인 공간 사용
