```python
num = input("숫자 입력")

# type(num)
# 사용자에게 숫자를 입력받은 후 +10 한 결과 출력
if num.isdigit():
    num = int(num) + 10
    print(num)
else:
    print("숫자를 입력하세요")
```

```python
def calculator(num1,num2,op):
    if op == "+":
        return num1 + num2
    elif op == "-":
        return num1 - num2
    elif op == "*":
        return num1 * num2
    elif op == "/":
        return num1 / num2
    elif op == "//":
        return num1 // num2

# 숫자 2개 입력받기, 연산자 기호(+,*...) 입력받기
num1 = int(input('press first number'))
num2 = int(input('presssecond number'))
op = input("press operator(+,-,*,/,//)")

print(calculator(num1,num2,op))
```

```python
values = input("두 숫자와 연산자 기호 입력 (ex 15 10 +)")

# 문자(공백 기준으로 분리)
# '15  10 +
# result = values.split()
# print(result)

num1,num2,op = values.split()

print(calculator(int(num1),int(num2),op))
```

```python
# 15,10,+

values = input("두 숫자와 연산기호입력")
num1,num2,op = values.split(",")
print(calculator(int(num1),int(num2),op))
```
