```python
def add_mul(choice, *args,):
    if choice == "add":
        result = 0
        for i in args:
            result = result + i
    elif choice == "mul":
        result = 1
        for i in args:
            result = result * i
    return result

print(add_mul("add",2,2))
print(add_mul("mul",2,2))
```

```python
# 딕셔너리 == 키워드 매개변수
def print_kwargs(**kwargs):
    print(kwargs)

    # key 값만 출력
    for k in kwargs.keys():
        print(k)

    # value 값만 출력
    for v in kwargs.values():
        print(v)

    # key, value 같이 출력
        for k,v in kwargs.items():
            print(k,v)

print(print_kwargs(a="qq", b="tt"))
```

```python
# Account 클래스 작성
# 멤버변수 : 계좌주(owner), 잔액(balance)
#   메소드 : deposit(money) : 잔액 = 잔액 + 입금액
#   메소드 : withdraw(money) : 잔액 = 잔액 - 출금액
#   메소드 : showBalance(money) : 현재 잔액 출력

class Account:
    
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, money):
        self.balance = self.balance + money

    def withdraw(self, money):
        if self.balance < money:
            print("잔액 부족")
        else:
            self.balance = self.balance - money

    def showBalance(self):
        print(f"현재 잔액 : {self.balance}")




a1 = Account("몽가",3000)
a2 = Account("사기",30000)

a1.deposit(2000)
a2.withdraw(200)
a1.showBalance()
a2.showBalance()
```

```python
# Audio : 멤버변수 : power, volume / 메소드 : 전원 조절, 볼륭 조절 + tune()
# TV : 멤버변수 : power, volume, size / 메소드 : 전원 조절, 볼륨 조절 + watch()
# 처음부터 설계하고 부모로 만들고 겹치지 않는 부분은 개별로 추가

class AudioVisual:
    def __init__(self, power, volume):
        self.power = power
        self.volume = volume


    def switch(self, on_off):
        self.power = on_off

    def setVolume(self, volume):
        self.volume = volume

class Audio(AudioVisual):
    def __init__(self, power, volume):
        super().__init__(power, volume)

    def tune(self):
        print("La la la..." if self.power else "turn it on")


class TV(AudioVisual):
    def __init__(self, power, volume, size):
        super().__init__(power, volume)
        self.size = size

    def watch(self):
        print("have fun!!" if self.power else "switch on")


tv = TV(False, 12, 50)
# tv 켜기
tv.switch(True)
# tv 보기
tv.watch()

audio = Audio(True, 15)
audio.setVolume(10)
audio.tune()
audio.switch(False)
```
