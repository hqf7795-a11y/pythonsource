````python
# add(2,3) or add(2,3,4,5) or add(2,3,4,5,6,7)

# 가변 매개변수
# 입력값이 몇 개가 될 지 모르는 경우
# *

# def 함수명(*변수명)
    # 해야할 일


def add(*args):
    print(args)
add(2,3)
add(2,3,4,5)
add(2,3,4,5,6,7)

def add(*args):
    result = 0
    for i in args:
        result = result + i
    print(result)

add(111,111,111,111,111)
````

````python
# def add_mul(choice, *args, a):
# args가 다 먹으니까 구분해주는거임

# 일반 매개변수는 앞 쪽으로 선언하는편이래
# +, * 실행
def add_mul(choice, *args,):
    if choice == "add":
        result = 0
        for i in args:
            result = result + i
    elif choice == "mul":
        result = 1
        for i in args:
            result = result * i

    return result

print(add_mul("add",2,2,2,2,2))
print(add_mul("mul",2,2,2,2))
````

````python
# 키워드 매개변수 (key=value)
# ** 변수명

# 딕셔너리 == 키워드 매개변수
def print_kwargs(**kwargs):
    print(kwargs)

    # key 값만 출력
    for k in kwargs.keys():
        print(k)

    # value 값만 출력
    for v in kwargs.values():
        print(v)

    # key, value 같이 출력
        for k,v in kwargs.items():
            print(k,v)
````

````python
# 키워드 매개변수 (key = value)
# **변수명

# 딕셔너리 == 키워드 매개변수
def print_kwargs(**kwargs):
    print(kwargs)
````

````python
# 변수의 효력 범위


# 함수 밖의 변수 : 전역변수
a = 1
def vartest(a):
    # 매개변수 a : 지역변수
    a = a + 1

vartest(a)
print(a)
````

````python
# 전역변수 a 값 변경하기
# 1) 리턴 이용 (추천)

a = 1
def vartest(a):
    # 매개변수 a : 지역변수
    a = a + 1
    return a

a = vartest(a)
print(a)
````

````python
# 2) global 키워드 사용 (비추)
# 함수 안에서 전역변수 접근 불가

a = 1
def vartest():
    global a
    a = a + 1

vartest()
print(a)
````

````python
# 입력으로 숫자(자연수)를 받아서 자연수가 짝수라며 True, 아니면 False 리턴
# is_even()

def is_even(X):
    if X %2 == 0:
        return True
    else:
        return False
    
print(is_even(22))
print(is_even(21))

# result = is_even(6)
# if result:
    # print("짝수")
# else:
    # print("홀수")

"짝수" if is_even(5) else "홀수"
````

````python
# 리스트의 평균을 구하는 함수
# get_average()
# 요소의 개수는 정해지지 않음
# no return



def get_everage(*args):
    result = 0
    for i in args:
        result = result + i

    print(result / len(args))

get_everage(11,22,33,44)
get_everage(11,22,55,44)
get_everage(11,22,77,44)
````

````python
#sum

def get_everage2(*args):
    print(sum(args) / len(args))

get_everage2(11,22,33,44)
get_everage2(11,22,55,44)
get_everage2(11,22,77,44)
````

````python
# 사칙연산 함수 use if elif result return

def calculator(num1,num2,op):
    if op == "+":
        return num1 + num2
    elif op == "-":
        return num1 - num2
    elif op == "*":
        return num1 * num2
    elif op == "/":
        return num1 / num2
    elif op == "//":
        return num1 // num2



print(calculator(10,20,"+"))
print(calculator(10,20,"-"))
print(calculator(10,20,"*"))
print(calculator(10,20,"/"))
print(calculator(10,20,"//"))
````

````python
# filter(함수 , iterable)

list1 = list(range(1,21))
list1

# 짝수만 새 리스트로 생성
# 방법 1

even_list = []
for i in list1:
    if i % 2 == 0:
        even_list.append(i)

# 방법2 (리스트 컴프리헨션)
even_list = [i for i in list1 if i % 2 == 0]

# 방법3 filter

list(filter(lambda x:x%2==0, list1))
````
