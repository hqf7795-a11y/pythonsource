```python
class Dog:
    # 클래스 변수
    species = "멍멍이" # 공통된 속성은 self 쓸 필요가없대
    count = 0

    def __init__(self,name,age,):
        self.name = name
        self.age = age
        Dog.count += 1

    def bark(self):
        return f"{self.name}:밥줘"

    @classmethod
    def get_count(cls):
        return f"지금까지 만들어진 {cls.species}는 {cls.count}입니다."

dog1 = Dog("john","10")
dog2 = Dog("man","11")

print(dog1.name, dog1.age)
print(dog2.name, dog1.age)

print(dog1.bark())
print(dog2.bark())

print(Dog.get_count())
```

```python
# Account 클래스 작성
# 멤버변수 : 계좌주(owner), 잔액(balance)
#   메소드 : deposit(money) : 잔액 = 잔액 + 입금액
#   메소드 : withdraw(money) : 잔액 = 잔액 - 출금액
#   메소드 : showBalance(money) : 현재 잔액 출력

class Account:
    
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, money):
        self.balance = self.balance + money

    def withdraw(self, money):
        if self.balance < money:
            print("잔액 부족")
        else:
            self.balance = self.balance - money

    def showBalance(self):
        print(f"현재 잔액 : {self.balance}")




a1 = Account("몽가",3000)
a2 = Account("사기",30000)

a1.deposit(2000)
a2.withdraw(200)
a1.showBalance()
a2.showBalance()
```

```python
# Order class
#  클래스 변수 : total_order
#  멤버변수: 메뉴(menu), 가격(price), 수량(count)
# 메소드 : getTotalPrice() = 가격 x 수량 : 총 주문금액 반환
# showOrder() : 출력문 작성
# 메뉴 : 아메리카노
# 가격 : 4000원
# 수량 :  2개
#  총 주문금액 : 8000원


class Order:
    total_order = 0

    def __init__(self,menu,price,count):
        self.menu = menu
        self.price = price
        self.count = count
        Order.total_order += 1

    def getTotalPrice(self):
        return self.price * self.count

    def showOrder(self):
        print(f"메뉴:{ self.menu}\n가격:{self.price}\n수량:{self.count}\n총 주문금액:{self.getTotalPrice()}")


order1 = Order("아아",1100,1)
order2 = Order("뜨아",1100,2)

order1.showOrder()
print()
order2.showOrder()

print(f"전체주문수 : {Order.total_order}")
```

#### 상속
- 기존 클래스(부모/상위)의 멤버변수와 메서드를 새로운 클래스(자식/하위)가 그대로 물려받는 것
- 코드 재사용성, 확장

```python
class Parent:
    def __init__(self):
        self.value = "테스트"
        print("Parent 클래스의 __init__() 호출")

    def test(self):
        print("Parent 클래스의 test() 호출")


class Child(Parent):
    pass

child = Child()
child.test()
print(child.value)
```

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def eat(self):
        print(f"{self.name}가 먹이를 먹는다")

    def sound(self):
        print(f"{self.name}이 소리를 낸다")


class Dog(Animal):
    def __init__(self, name, kind):
        super().__init__(name)
        self.kind = kind

    def fetch(self):
        print(f"{self.name}이 공을 물어옵니다.")

dog = Dog("john","mix")
dog.eat()
dog.sound()
dog.fetch()
```

```python
# Audio : 멤버변수 : power, volume / 메소드 : 전원 조절, 볼륭 조절 + tune()
# TV : 멤버변수 : power, volume, size / 메소드 : 전원 조절, 볼륨 조절 + watch()
# 처음부터 설계하고 부모로 만들고 겹치지 않는 부분은 개별로 추가

class AudioVisual:
    def __init__(self, power, volume):
        self.power = power
        self.volume = volume


    def switch(self, on_off):
        self.power = on_off

    def setVolume(self, volume):
        self.volume = volume

class Audio(AudioVisual):
    def __init__(self, power, volume):
        super().__init__(power, volume)

    def tune(self):
        print("La la la..." if self.power else "turn it on")


class TV(AudioVisual):
    def __init__(self, power, volume, size):
        super().__init__(power, volume)
        self.size = size

    def watch(self):
        print("have fun!!" if self.power else "switch on")


tv = TV(False, 12, 50)
# tv 켜기
tv.switch(True)
# tv 보기
tv.watch()

audio = Audio(True, 15)
audio.setVolume(10)
audio.tune()
audio.switch(False)
```

#### 오버라이딩
- 부모 클래스에 있는 메서드를 자식 클래스에서 같은 이름으로 다시 정의해서 동작내용 변경

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def sound(self):
        print(f"{self.name}가 소리를 냅니다.")

class Dog(Animal):
    def sound(self):
        # 오버라이딩시 동일한 부모 메소드를 호출하고싶다면
        super().sound()
        print(f"{self.name} : 멍멍!!")
        # return  super().sound()

class Cat(Animal):
    def sound(self):
        # super().sound()
        print(f"{self.name} : 야옹!!")
        # return super().sound()
    
dog = Dog("aa")
cat = Cat("ss")

dog.sound()
cat.sound()
```

```python
# Employee 부모 클래스 :
# 멤버 변수 : 이름(name), 급여(pay)
#   메소드 : showInfo() => 이름, 급여 출력

# Manager 자식 클래스
#  멤버 변수 : 보너스(bonus)
#   메소드 : getTotalPay() : 급여 + 보너스 출력


class Employee:
    def __init__(self,name,pay):
        self.name = name
        self.pay = pay

    def showInfo(self):
        print(f"{self.name}: {self.pay}")

class Manager(Employee):
    def __init__(self, name, pay, bonus): #여기에 추가함 잊지망
        super().__init__(name,pay)
        self.bonus = bonus

    def getTotalpay(self):
        print(f"{self.name}: {self.pay + self.bonus}")


zz = Employee("가",10)
zz.showInfo()
xx = Manager("나",20,20)
xx.showInfo()
xx.getTotalpay()
```

```python
# Shape 부모 클래스
# 메서드 : area() => "도형의 넓이" 출력

# Rectangle 자식클래스
# 멤버변수 : width, height
# 메서드 : area() => 직사각형의 넓이 출력

# Circle 자식 클래스
# 멤버 변수 : 반지름(radius)
# 메서드 : area() => 원의 넓이 출력

class Shape:
    
    def area(self):
        print(f"도형의 넓이")

class Rectangle(Shape):

    def __init__(self,width,height):
        super().__init__()
        self.width = width
        self.height = height

    def area(self):
        print(f"사각형 넓이: {self.width * self.height}")

class Circle(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius = radius

    def area(self):
        print(f"원의 넓이: {self.radius * self.radius * 3.14}")

r = Rectangle(10,20)
r.area()

c = Circle(5)
c.area()
```
